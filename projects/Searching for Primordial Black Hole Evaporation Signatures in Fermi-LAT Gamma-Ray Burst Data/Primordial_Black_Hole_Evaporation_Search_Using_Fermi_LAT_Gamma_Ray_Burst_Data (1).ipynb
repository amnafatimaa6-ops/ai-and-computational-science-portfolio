# Looking for Tiny-Black-Hole Candidates in Fermi Gamma-Ray Burst Data

## What is this project about?

Space telescopes sometimes catch sudden flashes of gamma rays (the most energetic kind of light). These flashes are called **gamma-ray bursts**, or GRBs. NASA's Fermi telescope has recorded hundreds of them.

This project starts from a theory: very small black holes, which may have formed long ago, slowly lose energy, and one with a mass of about **5 × 10¹¹ kg** is expected to finish this process around today's age of the Universe. The theory is called **Hawking radiation**, and a tiny black hole of this kind is called a **primordial black hole** (PBH).

I did not test this theory. I used it as my starting point and asked a simple question: **do any bursts in the real Fermi data have numbers that look like what this idea would lead us to expect?**

This project is **not** trying to prove that primordial black holes exist. It is a first, simple search that makes a ranked list of bursts that could be looked at more closely later.

## Objectives

5. Pick out the very short bursts (shorter than 2 seconds).

## Results

### The data

The Fermi-LAT catalog has **231 bursts** and **111 columns**. Most columns are complete. The main gaps are:

- **Redshift** is missing for 186 bursts, so only **45 bursts** have one.
- **Fluence** is missing for 12 bursts.

I kept 8 columns: name, duration (`gbm_cat_t90`), fluence, flux, spectral index, detection strength (`like_lat_ts`), highest photon energy (`ext_emission_max_ene`) and redshift.

### The Hawking temperature

Using the formula in the code, a 5 × 10¹¹ kg black hole comes out at **21.145 MeV**. This is a calculation from the formula, not a measurement from data.

### What the whole sample looks like

| Quantity | Smallest | Median | Mean | Largest |
|---|---|---|---|---|
| Duration, T90 (seconds) | 0.128 | 31.232 | 54.909 | 478.026 |
| Detection strength (TS) | 0 | 47.11 | 187.10 | 4089.38 |
| Highest photon energy (MeV) | -670 (placeholder) | 2100 | 5971.86 | 94000 |
| Spectral index | -5.99995 | -2.09 | -2.19 | 0 |

- The typical burst lasts about **31 seconds**, so short bursts are not the usual case in this data.
- An energy of **-670** cannot be real, so it is a placeholder for missing data. I handled it when building the score.
- I also made histograms of duration, highest photon energy and spectral index, and a heatmap showing how the numbers move together. I used them to explore the data, not to build the final ranking.

### The short bursts

**21 bursts** last less than 2 seconds, about **9%** of the 231. They range from 0.128 s to 1.728 s. Only one of them has a redshift: **GRB090510016** (0.903). The other 20 have none in the catalog.

### The first score (`PBH_score`)

This quick score gives more points for being shorter, for stronger detection and for having no redshift value, plus a few small extra points. Top 5:

|---|---|---|
| 1 | GRB170127067 | 35.59 |
| 2 | GRB200415367 | 31.39 |
| 3 | GRB160702516 | 24.39 |
| 4 | GRB190606080 | 22.65 |
| 5 | GRB090510016 | 22.08 |

### The PBH Consistency Index

**What the code does:**

1. Photon energies of zero or less are treated as missing.
2. Bursts missing any of the four numbers are left out. Two bursts (GRB090227772 and GRB110529034) had the placeholder energy, so **19 of the 21** short bursts were scored.
3. Duration is turned into 1 / duration, and the spectral index is multiplied by -1.
4. Every number is rescaled to a 0 to 1 range (min-max scaling).
5. The four numbers are added with these weights: **duration 40%, highest photon energy 30%, detection strength 20%, spectral index 10%**. I chose these weights myself.

**Top 5:**

| Rank | Burst | Score | Duration (s) | Detection strength (TS) | Highest photon energy (MeV) | Redshift |
|---|---|---|---|---|---|---|
| 1 | GRB090510016 | 0.568 | 0.960 | 2923.6 | 30000 | 0.903 |
| 2 | GRB170127067 | 0.506 | 0.128 | 35.9 | 510 | none |
| 3 | GRB200415367 | 0.391 | 0.144 | 30.6 | 1700 | none |
| 4 | GRB190606080 | 0.308 | 0.224 | 9.4 | 1600 | none |
| 5 | GRB160702516 | 0.297 | 0.200 | 19.3 | 4800 | none |

GRB090510016 has the highest score, but I did not count it as a candidate because it is the only short burst with a measured redshift (0.903). It stays in the table so the full ranking is visible.

```python
# Remove unnecessary columns that are not required for the analysis.
# 'errors="ignore"' prevents an error if any of these columns are missing.
df = df.drop(
    columns=[
        "__row",
        "__x_ra_dec",
        "__y_ra_dec",
        "__z_ra_dec"
    ],
    errors="ignore"
)

# Create a new DataFrame containing only the features
# relevant to the Primordial Black Hole (PBH) analysis.
pbh_df = df[
[
    "name",                  # Gamma-Ray Burst (GRB) identifier
    "gbm_cat_t90",           # Burst duration (T90) in seconds
    "gbm_cat_fluence",       # Total gamma-ray energy received (fluence)
    "like_best_flux",        # Best estimated gamma-ray flux
    "like_best_grbindex",    # Spectral index of the GRB
    "like_lat_ts",           # LAT Test Statistic (detection significance)
    "ext_emission_max_ene",  # Maximum energy of extended emission
    "redshift"               # Cosmological redshift (distance indicator)
]
].copy()
```

```python
# Select bursts shorter than 2 seconds
short_grbs = pbh_df[pbh_df["gbm_cat_t90"] < 2].copy()

print(f"Number of short GRBs: {len(short_grbs)}")

short_grbs[
    [
        "name",
        "gbm_cat_t90",
        "like_lat_ts",
        "ext_emission_max_ene",
        "like_best_grbindex",
        "redshift"
    ]
].sort_values("gbm_cat_t90")
```

```python
candidate_df = short_grbs.copy()

candidate_df["duration_score"] = 1 / candidate_df["gbm_cat_t90"]

# Reliable detections
candidate_df["ts_score"] = np.log1p(candidate_df["like_lat_ts"])

# Spectral index
candidate_df["spectral_score"] = -candidate_df["like_best_grbindex"]

# Unknown redshift receives a small bonus
candidate_df["redshift_score"] = np.where(
    candidate_df["redshift"].isna(),
    1,
    0
)

# Penalize events with invalid photon energies
candidate_df["energy_score"] = np.where(
    candidate_df["ext_emission_max_ene"] > 0,
    1,
    0
)

candidate_df["PBH_score"] = (
    3 * candidate_df["duration_score"] +
    2 * candidate_df["ts_score"] +
    candidate_df["spectral_score"] +
    candidate_df["redshift_score"] +
    candidate_df["energy_score"]
)

candidate_df = candidate_df.sort_values(
    "PBH_score",
    ascending=False
)

candidate_df[
[
    "name",
    "PBH_score",
    "gbm_cat_t90",
    "like_lat_ts",
    "ext_emission_max_ene",
    "like_best_grbindex",
    "redshift"
]
].head(10)
```

```python
# Make a copy
short_grbs = short_grbs.copy()


short_grbs.loc[
    short_grbs["ext_emission_max_ene"] <= 0,
    "ext_emission_max_ene"
] = np.nan

# Features used in the analysis
features = [
    "gbm_cat_t90",
    "like_lat_ts",
    "ext_emission_max_ene",
    "like_best_grbindex"
]

# Keep only rows with complete information
analysis = short_grbs[features].dropna().copy()

# Flip values so that larger numbers score higher
analysis["gbm_cat_t90"] = 1 / analysis["gbm_cat_t90"]          # shorter bursts score higher
analysis["like_best_grbindex"] = -analysis["like_best_grbindex"]  # harder spectra score higher

# Normalize all features between 0 and 1
scaler = MinMaxScaler()

analysis_scaled = pd.DataFrame(
    scaler.fit_transform(analysis),
    columns=features,
    index=analysis.index
)

weights = {
    "gbm_cat_t90": 0.40,
    "like_lat_ts": 0.20,
    "ext_emission_max_ene": 0.30,
    "like_best_grbindex": 0.10
}

# Compute weighted PBH Consistency Index
analysis_scaled["PBH_consistency"] = (
    weights["gbm_cat_t90"] * analysis_scaled["gbm_cat_t90"] +
    weights["like_lat_ts"] * analysis_scaled["like_lat_ts"] +
    weights["ext_emission_max_ene"] * analysis_scaled["ext_emission_max_ene"] +
    weights["like_best_grbindex"] * analysis_scaled["like_best_grbindex"]
)

# Save scores back into the original dataframe
short_grbs["PBH_consistency"] = np.nan
short_grbs.loc[
    analysis_scaled.index,
    "PBH_consistency"
] = analysis_scaled["PBH_consistency"]

# Show the highest-ranking PBH candidates
top_candidates = short_grbs.sort_values(
    "PBH_consistency",
    ascending=False
)

display(
    top_candidates[
        [
            "name",
            "PBH_consistency",
            "gbm_cat_t90",
            "like_lat_ts",
            "ext_emission_max_ene",
            "like_best_grbindex",
            "redshift"
        ]
    ].head(10)
)
```

###What I found

Most bursts are not short. The typical burst lasts about 31 seconds, and only 21 of 231 (about 9%) last under 2 seconds.

In the PBH Consistency Index, the top-scoring burst is GRB090510016, but it is the only short burst with a measured redshift, so I did not count it as a candidate.

Among the rest, the highest-ranked bursts are GRB170127067 and GRB200415367, followed by GRB190606080 and GRB160702516.

The first score also put GRB170127067, GRB200415367 and GRB160702516 in its top three.

###What this does not mean

The ranking compares numbers in a catalog. It does not show that any burst is a primordial black hole.

The weights in the score were chosen by me, and only 19 bursts were scored, so the order could change if the choices changed.
